$$\frac{d^2x}{dt^2} + \omega^2x = 0$$
$$v = \frac{dx}{dt}, \quad a = \frac{d^2x}{dt^2}$$
$$Loss_{physics} = \frac{1}{N} \sum_{i=1}^{N} \left( \frac{d^2x_i}{dt^2} + \omega^2 x_i \right)^2$$
$$Loss_{ic} = (x(0) - 1)^2 + \left(\frac{dx}{dt}\Big|_{t=0} - 0\right)^2$$
$$Loss_{total} = Loss_{physics} + Loss_{ic}$$
$$x_{exact}(t) = \cos(\omega t)$$

```python
import torch
import torch.nn as nn
import matplotlib.pyplot as plt
import numpy as np

# 1. 신경망 모델 정의 (입력: 시간 t, 출력: 위치 x)
class PINN(nn.Module):
    def __init__(self):
        super().__init__()
        # 1 -> 32 -> 32 -> 1 의 아주 간단한 인공신경망
        self.net = nn.Sequential(
            nn.Linear(1, 32),
            nn.Tanh(),
            nn.Linear(32, 32),
            nn.Tanh(),
            nn.Linear(32, 1)
        )

    def forward(self, t):
        return self.net(t)

model = PINN()
optimizer = torch.optim.Adam(model.parameters(), lr=0.01)

# 물리 상수 (각진동수)
omega = 2.0

# 2. 물리 법칙(손실 함수) 정의 및 학습
epochs = 2000
for epoch in range(epochs):
    optimizer.zero_grad()

    # 학습할 시간(t) 범위 생성 (0 ~ 2초 사이의 무작위 지점들)
    # requires_grad=True 가 핵심! (시간 t에 대해 미분하기 위해 필수)
    t_physics = torch.rand(100, 1) * 2.0
    t_physics.requires_grad = True

    # 네트워크를 통해 위치 x 예측
    x_pred = model(t_physics)

    # PyTorch의 자동미분(Autograd)으로 속도(v)와 가속도(a) 계산
    dx_dt = torch.autograd.grad(x_pred, t_physics, grad_outputs=torch.ones_like(x_pred), create_graph=True)[0]
    d2x_dt2 = torch.autograd.grad(dx_dt, t_physics, grad_outputs=torch.ones_like(dx_dt), create_graph=True)[0]

    # 물리 방정식 오차: (가속도 + w^2 * 위치)가 0에 가까워져야 함
    physics_loss = torch.mean((d2x_dt2 + omega**2 * x_pred)**2)

    # 초기 조건 오차: t=0일 때 위치 x=1, 속도 v=0 이라고 가정
    t_initial = torch.tensor([[0.0]], requires_grad=True)
    x_initial = model(t_initial)
    v_initial = torch.autograd.grad(x_initial, t_initial, create_graph=True)[0]

    ic_loss = (x_initial - 1.0)**2 + (v_initial - 0.0)**2

    # 총 오차 = 물리 오차 + 초기조건 오차
    total_loss = physics_loss + ic_loss

    # 역전파 및 가중치 업데이트 (AI가 물리 법칙을 깨닫는 과정)
    total_loss.backward()
    optimizer.step()

    if epoch % 500 == 0:
        print(f"Epoch {epoch}, Loss: {total_loss.item():.5f}")

# 3. 결과 시각화
t_test = torch.linspace(0, 2, 100).view(-1, 1)
x_test = model(t_test).detach().numpy() # AI가 예측한 궤적
x_exact = np.cos(omega * t_test.numpy()) # 실제 물리 정답 (cos 그래프)

plt.figure(figsize=(8, 4))
plt.plot(t_test.numpy(), x_exact, label="Exact (Physics)", color='blue')
plt.plot(t_test.numpy(), x_test, label="PINN (AI Prediction)", color='red', linestyle='--')
plt.legend()
plt.title("Simple Harmonic Oscillator solved by PINN")
plt.show()
```

```python
# 1. AI의 '뇌' 구조 정의 (시간 t를 넣으면 x, y 좌표 2개가 나옴)
class SimplePINN(nn.Module):
    def __init__(self,nodes=32,outputs=2):
        super().__init__()
        self.net = nn.Sequential(
            nn.Linear(1, nodes),
            nn.Tanh(),
            nn.Linear(nodes, nodes),
            nn.Tanh(),
            nn.Linear(nodes, outputs) # 출력: [x위치, y위치]
        )

    def forward(self, t):
        return self.net(t)

# 모델과 최적화 도구 설정
model = SimplePINN(64)
optimizer = torch.optim.Adam(model.parameters(), lr=0.005)

# 물리 상수 설정
g = 9.8      # 중력 가속도
x0, y0 = 20.0, 30.0   # 0초일 때 시작 위치
vx0, vy0 = 10.0, 10.0 # 0초일 때 시작 속도 (발사 속도)

#AI설정
mean_time = 5.0
target_loss = 0.1

# 2. 학습 시작 (AI에게 물리 법칙 가르치기)
epochs = 0
while True:
    optimizer.zero_grad()

    # [단계 1] 무작위 시간 데이터 생성 (0~5초 사이 200개)
    t_physics = torch.rand(200, 1) * mean_time
    t_physics.requires_grad = True # "이 시간을 기준으로 미분할 거야!" 설정

    # [단계 2] AI의 예측값 받아오기
    preds = model(t_physics)
    x = preds[:, 0:1]
    y = preds[:, 1:2]

    # [단계 3] 자동 미분으로 속도와 가속도 계산
    # x를 t로 미분 -> vx, vx를 t로 미분 -> ax
    vx = torch.autograd.grad(x, t_physics, torch.ones_like(x), create_graph=True)[0]
    ax = torch.autograd.grad(vx, t_physics, torch.ones_like(vx), create_graph=True)[0]
    
    # y를 t로 미분 -> vy, vy를 t로 미분 -> ay
    vy = torch.autograd.grad(y, t_physics, torch.ones_like(y), create_graph=True)[0]
    ay = torch.autograd.grad(vy, t_physics, torch.ones_like(vy), create_graph=True)[0]

    # [단계 4] 물리 법칙 벌점(Loss) 계산
    # 가로(x): 가속도가 0이어야 함 (등속 운동)
    loss_physics_x = torch.mean(ax**2)
    # 세로(y): 가속도 + 9.8 = 0 이어야 함 (중력 가속도 -9.8)
    loss_physics_y = torch.mean((ay + g)**2)

    # [단계 5] 초기 조건 벌점 (0초일 때 위치와 속도가 맞는가?)
    t_0 = torch.tensor([[0.0]], requires_grad=True)
    p_0 = model(t_0)
    x_0_pred, y_0_pred = p_0[:, 0:1], p_0[:, 1:2]
    
    vx_0_pred = torch.autograd.grad(x_0_pred, t_0, create_graph=True)[0]
    vy_0_pred = torch.autograd.grad(y_0_pred, t_0, create_graph=True)[0]

    loss_ic = (x_0_pred - x0)**2 + (y_0_pred - y0)**2 + \
              (vx_0_pred - vx0)**2 + (vy_0_pred - vy0)**2

    # [단계 6] 전체 벌점 합산 및 학습
    total_loss = loss_physics_x + loss_physics_y + torch.mean(loss_ic)
    total_loss.backward()
    optimizer.step()

    epochs += 1
    if epochs % 500 == 0:
        print(f"Epoch {epochs}, Loss: {total_loss.item():.5f}")

    if total_loss.item() < target_loss:
        print(f'Epoches = {epochs}')
        break

# 3. 결과 시뮬레이션 및 그래프 그리기
print("학습 완료! 그래프를 그립니다...")
t_test = torch.linspace(0, 5, 100).view(-1, 1)
with torch.no_grad():
    preds_test = model(t_test).numpy()

plt.figure(figsize=(10, 6))
plt.plot(preds_test[:, 0], preds_test[:, 1], label='AI Trajectory', color='blue', linewidth=2)
plt.axhline(0, color='black', linestyle='--') # 지면
plt.title("Projectile Motion Predicted by PINN")
plt.xlabel("Distance (x)")
plt.ylabel("Height (y)")
plt.legend()
plt.grid(True)
plt.show()
```
